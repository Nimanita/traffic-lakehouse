```python
from pyspark.sql.functions import *
from pyspark.sql.types import *

```

```python
gold_traffic = spark.table("dev_catalog.silver.traffic") \
 .withColumn("Load_Time", current_timestamp())

gold_traffic.write.mode("overwrite").saveAsTable("dev_catalog.gold.gold_traffic")

```

```python

gold_roads = spark.table("dev_catalog.silver.roads") \
.withColumn("Load_Time", current_timestamp())

gold_roads.write.mode("overwrite").saveAsTable("dev_catalog.gold.gold_roads")
```
